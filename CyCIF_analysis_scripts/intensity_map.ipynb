```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from shapely.geometry import Polygon, Point
from scipy.stats import mannwhitneyu  # nonparametric inside vs outside test

cells_df = df.copy()
roi_select = roi_df.copy()  
x_col, y_col = "X_centroid", "Y_centroid"


def polygon_from_all_points(s: str) -> Polygon:
    """Parse OMERO 'all_points' string to a Shapely polygon."""
    pts = []
    for tok in str(s).split():
        x_str, y_str = tok.split(",")
        pts.append((float(x_str), float(y_str)))

    # close polygon if needed, then buffer(0) to fix self-intersections
    if pts[0] != pts[-1]:
        pts.append(pts[0])
    return Polygon(pts).buffer(0)

def cells_in_polygon(cells_sub: pd.DataFrame, poly: Polygon,
                     xcol=x_col, ycol=y_col) -> pd.Series:
    """Boolean mask: True for rows whose (x,y) lies inside or on boundary."""
    poly = poly.buffer(0)
    return cells_sub.apply(lambda r: poly.covers(Point(float(r[xcol]), float(r[ycol]))), axis=1)


# Build polygons from ROI CSV 
# all_points column appears in omero ROI export .csv file. if the ROI is rectangle or circle, needs other scripts
if "all_points" not in roi_select.columns:
    raise ValueError("ROI DataFrame must contain an 'all_points' column for polygon ROIs.")

roi_select = roi_select.dropna(subset=["all_points"]).copy()
roi_select["__polygon__"] = roi_select["all_points"].apply(polygon_from_all_points)

# Choose a readable key for each ROI
if "Id" in roi_select.columns:
    roi_select["__key__"] = roi_select["Id"].astype(str)
elif "Name" in roi_select.columns:
    roi_select["__key__"] = roi_select["Name"].astype(str)

# ------------------------------
# Prep cells
# ------------------------------
cells = cells_df.copy()
cells = cells.dropna(subset=[x_col, y_col])

# ------------------------------
# Inside/outside assignment for a chosen ROI (or loop over all)
# ------------------------------
# For now: use the first polygon; change idx to use another
idx = 0
poly = roi_select.iloc[idx]["__polygon__"]
roi_key = roi_select.iloc[idx]["__key__"]

cells["in_roi"] = cells_in_polygon(cells, poly)

roi_to_cell_ids = {}

for _, roi in roi_select.iterrows():
    poly = roi["__polygon__"]
    key = roi["__key__"]

    # boolean mask for inside cells
    mask = cells_in_polygon(cells, poly)
    
    # collect the cell IDs
    sel_ids = cells.loc[mask, "CellID"].tolist()
    roi_to_cell_ids[key] = sel_ids

# Suppose you already have a list of cell IDs inside the ROI
inside_ids = roi_to_cell_ids[roi_key]   

# Pick the columns you want
cols_of_interest = ["Ki67 pos", "EGFR pos", "HER2 pos",
       "Ki67_HER2_doublepos", "Ki67_EGFR_doublepos"]

# Extract those rows and columns by cellID
new_df = cells_df.loc[cells_df["CellID"].isin(inside_ids), cols_of_interest]

print(new_df.head(20))

```

```python
"""
Simplified version — no conditionals/branches

Usage: fill in Config with your DataFrame, ROI CSV path, centroid columns, markers, etc.
"""

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from shapely.geometry import LineString, Point

# keeps any cell whose centroid falls within 10 pixels on either side of that line.
class Config:
    dataframe = df   # set to your EPN_data
    roi_path = "C:/Users/yiw588/HMS Dropbox/Yi-Chien Wu/Projects/P211_ependymoma_scimap/Python/LSP33564.ome.tif-histolines_1.csv"
    out_prefix = "C:/Users/yiw588/HMS Dropbox/Yi-Chien Wu/Projects/P211_ependymoma_scimap/python/distance/histoline1"
    x_col = "X_centroid"
    y_col= "Y_centroid"
    markers = ["Ki67","HER2","EGFR","CD44"]
    pixel_size_um = 0.65  # a pixel size of 0.65 µm.
    band_width_px= 77     # 100 um area width 


def parse_roi_points(roi_path: str):
    df = pd.read_csv(roi_path)
    points = [(float(x), float(y)) for x,y in (tok.split(",") for tok in df.loc[0, "all_points"].split())]
    return points


def compute_distance_table(cfg: Config):
    q = cfg.dataframe
    pts = parse_roi_points(cfg.roi_path)
    ls = LineString(pts)

    recs = []
    for idx, row in q.iterrows():
        x, y = float(row[cfg.x_col]), float(row[cfg.y_col])
        p = Point(x,y)
        if p.distance(ls) <= cfg.band_width_px:
            s = ls.project(p)
            rec = {"cell_index": idx, "x": x, "y": y,
                   "dist_along_px": s,
                   "dist_along_um": s*cfg.pixel_size_um}
            for m in cfg.markers:
                rec[m] = row[m]
            recs.append(rec)
    return pd.DataFrame(recs)


def run(cfg: Config):
    df = compute_distance_table(cfg)
    df.to_csv(f"{cfg.out_prefix}_cells_along_line.csv", index=False)
    for m in cfg.markers:
        plt.scatter(df["dist_along_um"], df[m], s=8, alpha=0.4)
        plt.xlabel("Distance along ROI (µm)"); plt.ylabel(m)
        plt.yscale("log", base=2)  
        plt.show()
        plt.close()
    return df


# Example usage:
cfg = Config()
cfg.dataframe = df
cfg.roi_path = "C:/Users/yiw588/HMS Dropbox/Yi-Chien Wu/Projects/P211_ependymoma/Python/LSP33564.ome.tif-histolines_1.csv"
cfg.out_prefix = "C:/Users/yiw588/HMS Dropbox/Yi-Chien Wu/Projects/P211_ependymoma/Python/distance/histoline1_cells_along_line.csv"
cfg.markers = ["Ki67","HER2","EGFR","CD44"]
run(cfg)
```
