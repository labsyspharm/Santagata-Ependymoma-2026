```python
# --- t-SNE from a pandas DataFrame ------------------------------------------
# Requirements: pandas, numpy, scikit-learn, matplotlib

import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
import matplotlib.pyplot as plt


# ------------------------------------------
# 1) Inputs you can customize
# ------------------------------------------
# Your filtered dataframe with cells as rows
df_in = df_qc.copy()
df_in_sub50000 = df_in.sample(n=50000, random_state=42)

# Columns to use for embedding (numeric marker/features only)
# Tip: include only the features you want t-SNE to consider
feature_cols = [
    "EGFR", "HER2", "CD44", "SOX2", "FOXJ1_sigma"
]

# Optional colors (one categorical and one continuous example)
cat_col = None               # set to None if you don't have one
cont_col = "Ki67"            # set to None if you don't want a continuous overlay

# t-SNE parameters (feel free to tweak)
PERPLEXITY   = 30            # good range is ~5–50; smaller datasets -> smaller values
Niter= 1000                   # increase to 2000+ for a bit more stability
RANDOM_STATE = 0

# ------------------------------------------
# 2) Build the feature matrix
# ------------------------------------------
X = df_in_sub50000[feature_cols].copy()

# ---- NEW: log2 transform (add +1 to avoid log2(0)) ----
X = np.log2(X + 1)

# Drop rows with NaN/Inf values
mask_good = np.isfinite(X).all(axis=1)
X = X.loc[mask_good]
df_work = df_in_sub50000.loc[mask_good].copy()

# ------------------------------------------
# 3) Scale features (important for t-SNE)
# ------------------------------------------
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X.values)

# # (Optional) Light PCA before t-SNE for denoising/speed
# # Use enough PCs to explain most variance (e.g., 25 or min(n_features, 50))
n_pcs = min(25, X_scaled.shape[1])
X_pca = PCA(n_components=n_pcs, random_state=RANDOM_STATE).fit_transform(X_scaled)

# ------------------------------------------
# 4) Run t-SNE
# ------------------------------------------
tsne = TSNE(
    n_components=2,               # reduce data to 2 dimension 
    perplexity=PERPLEXITY,      # how many neighbors each point considers 
    early_exaggeration=12.0,      # multiplier that inflates the attractive forces
    max_iter=1000,
    learning_rate="auto",
    init="pca",                   # initial guess of where to put the points, init="random" or init="pca"
    method="barnes_hut",
    angle=0.5,
    random_state=RANDOM_STATE,
    verbose=1
)
X_tsne = tsne.fit_transform(X_pca)

# Save into dataframe
df_work["tSNE_1"] = X_tsne[:, 0]
df_work["tSNE_2"] = X_tsne[:, 1]    # i dont need to rerun everytime, I can use df_work for plotting and phenotyping 

# 5) Plotting
# ------------------------------------------
plt.figure(figsize=(6, 5))
plt.scatter(df_work["tSNE_1"], df_work["tSNE_2"], s=5)
plt.title("t-SNE (all cells)")
plt.xlabel("tSNE_1")
plt.ylabel("tSNE_2")
plt.tight_layout()
plt.show()
```

```python
# run elbow method to determin the value of K-mean clusters 

from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

inertia = []
K = range(2, 20)
for k in K:
    kmeans = KMeans(n_clusters=k, random_state=42).fit(X_scaled)
    inertia.append(kmeans.inertia_)

plt.plot(K, inertia, "o-")
plt.xlabel("k")
plt.ylabel("Inertia (within-cluster sum of squares)")
plt.title("Elbow method for optimal k")
plt.show()
```
